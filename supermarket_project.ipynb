## ⚙️ Milestone 2: Python Core, Custom Exception, and OOP Concepts
---



In this section, we process the supermarket dataset to demonstrate core Python skills and Object-Oriented Programming (OOP):
* **Python Core:** Process the first 20 rows using a custom function, list comprehension, map, filter, lambda, and sorting.
* **Custom Exception:** Implement `InvalidTransactionError` to validate price, quantity, and rating.
* **OOP Design:** Use `SaleTransaction` as a base class with `NormalSale` and `MemberSale` as derived classes to demonstrate encapsulation, inheritance, and polymorphism (rows 10 to 20).

```python
# ==========================================
# 1. Python Core Skills (First 20 Rows)
# ==========================================
# Take the first 20 rows of the DataFrame
sales_20 = df.head(20)

# Convert 'Sales' column to a list
sales_list = sales_20['Sales'].tolist()
# Convert 'Customer type' column to a list
customer_types = sales_20['Customer type'].tolist()

# A. Custom Function
def is_high_sale(amount):
    # Check if sale amount is greater than 100
    return amount > 100

# B. List Comprehension (Filter sales > 100)
high_sales = [s for s in sales_list if is_high_sale(s)]

# C. Filter & Lambda (Select only 'Member' customer types)
members_only = list(filter(lambda c: c == 'Member', customer_types))

# D. Map & Lambda (Add 10% tax to sales)
sales_with_tax = list(map(lambda s: round(s * 1.10, 2), sales_list))

# E. Sorting (Sort sales from lowest to highest)
sorted_sales = sorted(sales_list)

print("=== 🎯 Python Core Results ===")
print("High Sales (>100):", high_sales[:3])
print("Total Members in First 20 Rows:", len(members_only))
print("Sorted Sales (Lowest 3):", sorted_sales[:3])
print("\n")


# ==========================================
# 2. Custom Exception
# ==========================================
class InvalidTransactionError(Exception):
    """Custom exception for invalid transaction data"""
    def __init__(self, message):
        self.message = message
        super().__init__(self.message)


# ==========================================
# 3. Object-Oriented Programming (OOP)
# ==========================================
class SaleTransaction:
    """Base class for a sales transaction"""
    def __init__(self, invoice_id, price, quantity, rating=5):
        # Validate price, quantity, and rating
        if price <= 0 or quantity <= 0:
            raise InvalidTransactionError("Price and quantity must be positive!")
        if rating < 0 or rating > 10:
            raise InvalidTransactionError("Rating must be between 0 and 10!")

        self.invoice_id = invoice_id
        self._price = price        # Protected attribute for unit price (Encapsulation)
        self._quantity = quantity  # Protected attribute for quantity (Encapsulation)
        self.rating = rating

    def calculate_total(self):
        # Method to calculate total sale amount
        return self._price * self._quantity


class NormalSale(SaleTransaction):
    """Subclass for normal customer sales (no discount)"""
    def calculate_total(self):
        # Polymorphism: Full price with no discount
        return round(self._price * self._quantity, 2)


class MemberSale(SaleTransaction):
    """Subclass for member customer sales (5% discount)"""
    def calculate_total(self):
        # Polymorphism: Apply a 5% discount for members
        subtotal = self._price * self._quantity
        return round(subtotal * 0.95, 2)


# ==========================================
# 4. Processing Rows 10 to 20
# ==========================================
print("=== 🛍️ Processing Transactions (Rows 10 to 20) ===")

# Apply classes to rows 10 to 20 as requested
for index, row in sales_20.iloc[10:20].iterrows():
    try:
        inv_id = row['Invoice ID']
        u_price = row['Unit price']
        qty = row['Quantity']
        c_type = row['Customer type']
        rat = row['Rating'] if 'Rating' in row else 5

        # Create object based on customer type
        if c_type == 'Member':
            sale = MemberSale(inv_id, u_price, qty, rat)
        else:
            sale = NormalSale(inv_id, u_price, qty, rat)

        print(f"Invoice: {inv_id} | Type: {c_type:<6} | Final Total: ${sale.calculate_total()}")

    except InvalidTransactionError as e:
        # Catch and print custom validation errors
        print(f"⚠️ Error in Invoice {inv_id}: {e.message}")
```

## 🔢 Milestone 3: NumPy Data Analysis

In this section, we process numerical columns using **NumPy arrays** to perform fast computations without traditional loops:
* Convert `Unit price` and `Quantity` columns to NumPy arrays.
* Apply vectorized multiplication to calculate total sales.
* Use Boolean indexing to filter sales greater than $300.
* Calculate statistical measures (Mean, Max, Min, Std) for sales data.

```python
import numpy as np

# ==========================================
# 1. Convert Columns to NumPy Arrays
# ==========================================
# Convert 'Unit price' column to a NumPy array
prices = df['Unit price'].to_numpy()
# Convert 'Quantity' column to a NumPy array
quantities = df['Quantity'].to_numpy()


# ==========================================
# 2. Vectorized Calculation (No Loops)
# ==========================================
# Calculate total for each transaction by multiplying arrays
calculated_totals = prices * quantities


# ==========================================
# 3. Boolean Filtering Condition
# ==========================================
# Create a boolean mask for sales greater than $300
high_sales_mask = calculated_totals > 300
# Filter total sales using the boolean mask
high_sales_values = calculated_totals[high_sales_mask]


# ==========================================
# 4. Statistical Analysis (Mean, Max, Min, Std)
# ==========================================
# Calculate the mean (average) of all sales
mean_sale = np.mean(calculated_totals)
# Find the maximum sale value
max_sale = np.max(calculated_totals)
# Find the minimum sale value
min_sale = np.min(calculated_totals)
# Calculate the standard deviation of sales
std_sale = np.std(calculated_totals)

# Print the results nicely
print("=== 📊 NumPy Analysis Results ===")
print(f"Sample Calculated Totals (First 5): {np.round(calculated_totals[:5], 2)}")
print(f"Number of High Sales (> $300): {len(high_sales_values)}")
print("-" * 35)
print(f"📈 Sales Statistics:")
print(f" - Mean (Average): ${mean_sale:.2f}")
print(f" - Max  (Highest Value): ${max_sale:.2f}")
print(f" - Min  (Lowest Value): ${min_sale:.2f}")
print(f" - Std  (Standard Deviation): ${std_sale:.2f}")
```

## 🐼 Milestone 4: Pandas Analysis and Excel Export

In this final section, we perform data cleaning, answer business questions, create a pivot table, and export everything into an Excel workbook:
* **Data Cleaning:** Clean column names, remove duplicates, handle missing values, and convert the `Date` column to `datetime`.
* **Analysis Questions:** Answer 3 business questions using Pandas aggregations.
* **Pivot Table:** Summarize total sales by Branch and Payment Method.
* **Excel Export:** Save the cleaned data, pivot table, and analysis summary into `Supermarket_Sales_Report.xlsx`.

```python
import pandas as pd

# ==========================================
# 1. Data Cleaning
# ==========================================
# Create a copy of the DataFrame for cleaning
df_cleaned = df.copy()

# A. Clean column names (remove extra spaces)
df_cleaned.columns = df_cleaned.columns.str.strip()

# B. Handle duplicate rows (remove them)
df_cleaned = df_cleaned.drop_duplicates()

# C. Handle missing values (remove rows with any missing data)
df_cleaned = df_cleaned.dropna()

# D. Convert 'Date' column to datetime objects
df_cleaned['Date'] = pd.to_datetime(df_cleaned['Date'])


# ==========================================
# 2. Answer 3 Simple Analysis Questions
# ==========================================
# Q1: Total Sales by Branch
q1_branch_sales = df_cleaned.groupby('Branch')['Sales'].sum().reset_index()

# Q2: Total Quantity Sold by Product Line
q2_product_qty = df_cleaned.groupby('Product line')['Quantity'].sum().reset_index()

# Q3: Average Customer Rating by Customer Type
q3_customer_rating = df_cleaned.groupby('Customer type')['Rating'].mean().reset_index() if 'Rating' in df_cleaned.columns else pd.DataFrame()

print("=== 📊 Analysis Question 1: Total Sales by Branch ===")
print(q1_branch_sales)
print("\n=== 🛍️ Analysis Question 2: Total Quantity by Product Line ===")
print(q2_product_qty.head())


# ==========================================
# 3. Create One Pivot Table
# ==========================================
# Pivot table showing total sales for each branch, broken down by payment method
pivot_sales = pd.pivot_table(
    df_cleaned,
    values='Sales',
    index='Branch',
    columns='Payment',
    aggfunc='sum'
).reset_index()

print("\n=== 🧩 Pivot Table: Sales by Branch & Payment ===")
print(pivot_sales)


# ==========================================
# 4. Export to Excel File (Multiple Sheets)
# ==========================================
file_name = "Supermarket_Sales_Report.xlsx"

# Use ExcelWriter to export multiple sheets into the same file
with pd.ExcelWriter(file_name, engine='openpyxl') as writer:
    # 1. First Sheet: Cleaned Data
    df_cleaned.to_excel(writer, sheet_name='Cleaned Data', index=False)

    # 2. Second Sheet: Pivot Table
    pivot_sales.to_excel(writer, sheet_name='Pivot Table', index=False)

    # 3. Third Sheet: Analysis Summary and the three questions' answers
    q1_branch_sales.to_excel(writer, sheet_name='Summary Analysis', index=False, startrow=0)
    q2_product_qty.to_excel(writer, sheet_name='Summary Analysis', index=False, startrow=len(q1_branch_sales) + 3)
    # q3_customer_rating.to_excel(writer, sheet_name='Summary Analysis', index=False, startrow=len(q1_branch_sales) + len(q2_product_qty) + 6)

print(f"\n✅ Excel Report Successfully Created and Saved as '{file_name}'!")
```
